```python
from plotly import graph_objects as go
import pandas as pd
import matplotlib
```

```python
raw_data_eval_err=pd.read_csv("14_4/training_convergence_14bit.csv")
# raw_data_train_err=pd.read_csv("28/training_errors.csv")
```

```python
tot_data=25
# raw_data_eval_err
r_step=[i for i in range(tot_data)]
eval_errs=[]
for idx in range(tot_data):
    eval_errs.append(float(raw_data_eval_err["Mean_Reward"][idx]))
```

```python
fig=go.Figure()
fig.add_trace(go.Scatter(x=r_step, y=eval_errs))
fig.show()
```
